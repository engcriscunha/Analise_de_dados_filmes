### Distribuição Amostral e intervalo de confiança

```python
salarios = df['SALARIO']
salarios
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
media_amostral = np.mean(salarios)
media_amostral
```

```python
# desvio padrão amostral
desvio_amostral = np.std(salarios)
desvio_amostral
```

```python
#nivel confiança

nivel_confianca = 0.95
```

```python
tamanho_amostra = len(salarios)
tamanho_amostra
```

```python
from scipy import stats
```

```python
erro_padrao = stats.sem(salarios) #erro padrão da média (standard error of the mean)
erro_padrao
```

```python
intervalo_confianca =stats.t.interval(nivel_confianca, tamanho_amostra-1, loc=media_amostral, scale=erro_padrao)
intervalo_confianca
```

### Correlação, diferentes funções para dados discretos e contínuos

```python
correlacao_continua = df['IDADE'].corr(df['SALARIO'])
correlacao_continua
```

```python
from scipy.stats import chi2_contingency
```

```python
def cramer_coeficiente(coluna1, coluna2):
    tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
    chi2 = chi2_contingency(tabela_cruzada)[0]
    soma = np.sum(tabela_cruzada)
    mini = min(tabela_cruzada.shape) - 1
    cramer = np.sqrt(chi2 / (soma * mini))
    return cramer
```

```python
tabela_cruzada = np.array(pd.crosstab(df['COR/RACA/ETNIA'], df['NIVEL DE ENSINO']))
tabela_cruzada
```
